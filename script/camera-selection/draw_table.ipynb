```python
import matplotlib.pyplot as plt

x = [5, 8, 10, 15, 20]
x = [_ * 3 for _ in x]

latency = [60, 75, 85, 120, 165]
clip = [10.8, 11.0, 11.3, 13.5, 14.0]

fig, ax1 = plt.subplots()

# 각 포인트에 대해 점선 추가
for xi, yi in zip(x, latency):
    # 세로 점선 (x축에 평행)
    ax1.axhline(y=yi, color='gray', linestyle='--', alpha=0.4)
    # 가로 점선 (y축에 평행)
    ax1.axvline(x=xi, color='gray', linestyle='--', alpha=0.4)

ax1.plot(x, latency, marker='o')
ax1.set_xlabel("Edited View Number")
ax1.set_ylabel("Latency (s)")

ax2 = ax1.twinx()
ax2.scatter(x, clip, color='red', marker='x')
ax2.set_ylabel("CLIP Dir. Consistency", color='red')

plt.savefig("graph.pdf")  # 또는 "graph.svg"
```
